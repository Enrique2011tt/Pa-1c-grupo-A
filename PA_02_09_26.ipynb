```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]



soma_notas = sum(notas)
print(f"Soma das notas: {soma_notas}")


media_notas = soma_notas / len(notas)
print(f"Média das notas: {media_notas:.2f}")


maior_nota = max(notas)
print(f"Maior nota: {maior_nota}")


menor_nota = min(notas)
print(f"Menor nota: {menor_nota}")


if media_notas >= 7:
    print("Situação: Aprovado")
else:
    print("Situação: Reprovado")
```

```python
numeros = [2, 5, 2, 8, 2, 10, 5, 3, 2, 7]

contagem_2 = numeros.count(2)
print(f"O número 2 aparece {contagem_2} vezes na lista.")

contagem_5 = numeros.count(5)
print(f"O número 5 aparece {contagem_5} vezes na lista.")

se_10_presente = 10 in numeros
if se_10_presente:
    print("O número 10 está presente na lista.")
else:
    print("O número 10 NÃO está presente na lista.")

numero_usuario_str = input("Digite um número para verificar quantas vezes ele aparece na lista: ")
try:
    numero_usuario = int(numero_usuario_str)
    contagem_usuario = numeros.count(numero_usuario)
    print(f"O número {numero_usuario} aparece {contagem_usuario} vezes na lista.")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
numeros_originais = [-5, 10, -3, 8, 0, -1, 7, -9, 4]


numeros_positivos = []
numeros_negativos = []
numeros_zeros = []

for numero in numeros_originais:
    if numero > 0:
        numeros_positivos.append(numero)
    elif numero < 0:
        numeros_negativos.append(numero)
    else:
        numeros_zeros.append(numero)


print(f"Lista original: {numeros_originais}")
print(f"Números Positivos: {numeros_positivos}")
print(f"Números Negativos: {numeros_negativos}")
print(f"Números Zero: {numeros_zeros}")
```

```python
produtos = [
    ["Arroz", 25.90],
    ["Feijão", 8.50],
    ["Macarrão", 6.90],
    ["Leite", 5.50],
    ["Café", 15.00]
]


print("--- Lista de Produtos ---")
for nome, preco in produtos:
    print(f"{nome}: R$ {preco:.2f}")


valor_total_compra = 0
for nome, preco in produtos:
    valor_total_compra += preco
print(f"\nValor Total da Compra: R$ {valor_total_compra:.2f}")


produto_mais_caro = produtos[0]
produto_mais_barato = produtos[0]

for produto in produtos:
    if produto[1] > produto_mais_caro[1]:
        produto_mais_caro = produto
    if produto[1] < produto_mais_barato[1]:
        produto_mais_barato = produto

print(f"Produto Mais Caro: {produto_mais_caro[0]} (R$ {produto_mais_caro[1]:.2f})")
print(f"Produto Mais Barato: {produto_mais_barato[0]} (R$ {produto_mais_barato[1]:.2f})")


produto_procurado = input("\nDigite o nome de um produto para verificar se está cadastrado: ")
produto_encontrado = False

for nome, preco in produtos:
    if nome.lower() == produto_procurado.lower():
        print(f"O produto '{nome}' está cadastrado e custa R$ {preco:.2f}.")
        produto_encontrado = True
        break

if not produto_encontrado:
    print(f"O produto '{produto_procurado}' NÃO está cadastrado.")
```

```python
alunos_e_notas = []


num_alunos = int(input("Quantos alunos você deseja cadastrar? "))


for i in range(num_alunos):
    print(f"\n--- Cadastro do Aluno {i+1} ---")
    nome = input("Digite o nome do aluno: ")
    while True:
        try:
            nota = float(input("Digite a nota do aluno (0 a 10): "))
            if 0 <= nota <= 10:
                break
            else:
                print("Por favor, digite uma nota entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número para a nota.")


    alunos_e_notas.append({"nome": nome, "nota": nota})

print("\n--- Resultados ---")


print("\nAlunos e Suas Notas:")
for aluno in alunos_e_notas:
    print(f"  {aluno['nome']}: {aluno['nota']:.2f}")


if alunos_e_notas:
    soma_notas_turma = sum(aluno['nota'] for aluno in alunos_e_notas)
    media_turma = soma_notas_turma / len(alunos_e_notas)
    print(f"\nMédia da Turma: {media_turma:.2f}")

    aluno_maior_nota = max(alunos_e_notas, key=lambda x: x['nota'])
    print(f"Aluno com a Maior Nota: {aluno_maior_nota['nome']} ({aluno_maior_nota['nota']:.2f})")


    aluno_menor_nota = min(alunos_e_notas, key=lambda x: x['nota'])
    print(f"Aluno com a Menor Nota: {aluno_menor_nota['nome']} ({aluno_menor_nota['nota']:.2f})")


    alunos_aprovados = [aluno for aluno in alunos_e_notas if aluno['nota'] >= 7]
    print("\nAlunos Aprovados:")
    if alunos_aprovados:
        for aluno in alunos_aprovados:
            print(f"  {aluno['nome']} ({aluno['nota']:.2f})")
    else:
        print("  Nenhum aluno aprovado.")

    alunos_reprovados = [aluno for aluno in alunos_e_notas if aluno['nota'] < 7]
    print("\nAlunos Reprovados:")
    if alunos_reprovados:
        for aluno in alunos_reprovados:
            print(f"  {aluno['nome']} ({aluno['nota']:.2f})")
    else:
        print("  Nenhum aluno reprovado.")

else:
    print("Nenhum aluno cadastrado para exibir resultados.")

print("\n--- Pesquisa de Aluno ---")
if alunos_e_notas:
    nome_pesquisar = input("Digite o nome do aluno que deseja pesquisar: ")
    aluno_encontrado = False
    for aluno in alunos_e_notas:
        if aluno['nome'].lower() == nome_pesquisar.lower():
            print(f"A nota de {aluno['nome']} é {aluno['nota']:.2f}.")
            aluno_encontrado = True
            break
    if not aluno_encontrado:
        print(f"Aluno '{nome_pesquisar}' não encontrado.")
else:
    print("Não há alunos cadastrados para pesquisar.")
```

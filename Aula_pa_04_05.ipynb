```python
#escrever um programa que o usuario digita um numero e o programa imprime de 0 ate o numero
numero = int(input("Digite um número: "))

for i in range(numero + 1):
    print(i)
```

```python
#escrever um programa que o usuario digita um numero e o programa imprime em ordem decrescente ate 0
numero = int(input("Digite um número: "))

for i in range(numero, -1, -1):
    print(i)
```

```python
#peça um numero e mostre todos os numeros pares ate o numero
numero = int(input("Digite um número: "))

for i in range(numero + 1):
    if i % 2 == 0:
        print(i)
```

```python
#peça um numero e calcule a soma de 1 ate o numero
numero = int(input("Digite um número: "))

soma = 0
for i in range(1, numero + 1):
    soma += i
print(f"A soma de 1 até {numero} é: {soma}")
```

```python
#peça uma palavra e conte as letras ex python a palavra tem 6 letras
palavra = input("Digite uma palavra: ")
numero_letras = len(palavra)
print(f"{palavra} a palavra tem {numero_letras} letras")
```

```python
#peça uma palavra e conte as vogais
palavra = input("Digite uma palavra: ")
vogais = "aeiouAEIOU"
contador_vogais = 0

for letra in palavra:
    if letra in vogais:
        contador_vogais += 1

print(f"A palavra '{palavra}' tem {contador_vogais} vogais.")
```

```python
# 2- Soma dos Números
num_soma = int(input("Digite um número para somar até ele: "))
soma_total = 0
for i in range(1, num_soma + 1):
  soma_total += i
print(f"A soma dos números até {num_soma} é {soma_total}")
```

```python
# 3- Tabuada
A = int(input("digite um numero: "))
for i in range (1, 11):
  print (f"{A} X {i} = {A * i} ")
```

```python
# 4- Fatorial
num = int(input("Digite um número: "))
fatorial = 1

if num < 0:
    print("O fatorial não está definido para números negativos.")
elif num == 0:
    print("Digite outro número.")
else:
 for i in range(1, num + 1):
    fatorial *= i

print("O fatorial de", num, "é:", fatorial)
```

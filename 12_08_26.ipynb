```python
#4
#crie uma lista com 10 números inteiros
#percorra a lista e mostre quais números são pares ou impares
#ex : 4 é par

num = [1, 4, 10, 20, 23, 73, 33, 54, 67, 19]
for n in num:
    if n % 2 == 0:
        print(f"{n} é par")
    else:
        print(f"{n} é ímpar")
```

```python
#6
# considere:

notas = [7.5,8.0,6.5,9.0,7.0]

maior_nota = max(notas)
print(f"A maior nota é: {maior_nota}")


menor_nota = min(notas)
print(f"A menor nota é: {menor_nota}")


media_notas = sum(notas) / len(notas)
print(f"A média das notas é: {media_notas}")
```

```python
#7
# crie uma lista vazia chamada
# mostre os produtos númerados
# peça ao usuário para cadastrar 5 produtos utilizando for
produtos = []


for i in range(5):
    produto = input(f"Digite o nome do produto {i+1}: ")
    produtos.append(produto)


print("\nProdutos cadastrados:")
for i, produto in enumerate(produtos):
    print(f"#{i+1} {produto}")
```

```python
#8
#considere:
alunos = ["Ana","Carlos","Maria","Pedro","João"]

#peça ao usuário para digitar um nome, o programa deve verificar se o nome está ou não na lista
nome_digitado = input("Digite um nome para verificar na lista: ")

#faça o programa funcionar independente da forma que o usuário digitar ex: MARIA, Maria ou maria
nome_digitado_lower = nome_digitado.lower()
alunos_lower = [aluno.lower() for aluno in alunos]

if nome_digitado_lower in alunos_lower:
    print(f"O nome '{nome_digitado}' está na lista de alunos.")
else:
    print(f"O nome '{nome_digitado}' NÃO está na lista de alunos.")
```

```python
#9
#crie uma lista vazia chamada números
#peça ao usuário para digitar 10 números
#depois crie duas listas chamadas pares e impares
#percorre a lista original e coloque cada número na lista correspondente

numeros = []


for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    numeros.append(numero)


    pares = []
    impares = []


    for numero in numeros:
        if numero % 2 == 0:
            pares.append(numero)
        else:
            impares.append(numero)

print("Lista original:", numeros)
print("Números pares:", pares)
print("Números ímpares:", impares)
```

```python
#10
# faça um pequeno sistema para cadastrar a nota de 5 alunos
notas_alunos = []

print("Por favor, digite as notas de 5 alunos:")
for i in range(5):
    while True:
        try:
            nota = float(input(f"Digite a nota do aluno {i+1}: "))
            if 0 <= nota <= 10: # Assuming notes are between 0 and 10
                notas_alunos.append(nota)
                break
            else:
                print("A nota deve ser entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")

# o programa deverá armazenar as notas em uma lista e depois mostrar:

# todas as notas
print(f"\nTodas as notas: {notas_alunos}")

# maior nota
maior_nota_turma = max(notas_alunos)
print(f"Maior nota da turma: {maior_nota_turma}")

# média da turma
media_turma = sum(notas_alunos) / len(notas_alunos)
print(f"Média da turma: {media_turma:.2f}")

# quantidade de alunos com a nota maior ou igual que 6
alunos_aprovados = 0
for nota in notas_alunos:
    if nota >= 6:
        alunos_aprovados += 1
print(f"Quantidade de alunos com nota maior ou igual a 6: {alunos_aprovados}")

# quantidade de alunos com a nota menor que 6
alunos_reprovados = 0
for nota in notas_alunos:
    if nota < 6:
        alunos_reprovados += 1
print(f"Quantidade de alunos com nota menor que 6: {alunos_reprovados}")
```

```python
#4
#crie uma lista com 10 números inteiros
#percorra a lista e mostre quais números são pares ou impares
#ex : 4 é par

num = [1, 4, 10, 20, 23, 73, 33, 54, 67, 19]
for n in num:
    if n % 2 == 0:
        print(f"{n} é par")
    else:
        print(f"{n} é ímpar")
```

```python
#6
# considere:

notas = [7.5,8.0,6.5,9.0,7.0]

maior_nota = max(notas)
print(f"A maior nota é: {maior_nota}")


menor_nota = min(notas)
print(f"A menor nota é: {menor_nota}")


media_notas = sum(notas) / len(notas)
print(f"A média das notas é: {media_notas}")
```

```python
#7
# crie uma lista vazia chamada
# mostre os produtos númerados
# peça ao usuário para cadastrar 5 produtos utilizando for
produtos = []


for i in range(5):
    produto = input(f"Digite o nome do produto {i+1}: ")
    produtos.append(produto)


print("\nProdutos cadastrados:")
for i, produto in enumerate(produtos):
    print(f"#{i+1} {produto}")
```

```python
#8
#considere:
alunos = ["Ana","Carlos","Maria","Pedro","João"]

#peça ao usuário para digitar um nome, o programa deve verificar se o nome está ou não na lista
nome_digitado = input("Digite um nome para verificar na lista: ")

#faça o programa funcionar independente da forma que o usuário digitar ex: MARIA, Maria ou maria
nome_digitado_lower = nome_digitado.lower()
alunos_lower = [aluno.lower() for aluno in alunos]

if nome_digitado_lower in alunos_lower:
    print(f"O nome '{nome_digitado}' está na lista de alunos.")
else:
    print(f"O nome '{nome_digitado}' NÃO está na lista de alunos.")
```

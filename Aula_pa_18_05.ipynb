```python
# 1- Escrever um programa que o usuário digita um número de 1 a 10 e o programa imprimi a tabuada do número digitado
A = int(input("digite um numero: "))
for i in range (1, 11):
  print (f"{A} X {i} = {A * i} ")
```

```python
# 2- Escrever um programa que o usuário digita ate o número e o programa imprimi até o número digitado
num = int(input("Digite um número: "))
for i in range(1, num + 1):
  print(i)
```

```python
# 3- Escrever um programa que o usuário digita um número e o programa imprimi o número digita de forma regressiva
num_regressivo = int(input("Digite um número para contagem regressiva: "))
for i in range(num_regressivo, 0, -1):
  print(i)
```

```python
# 4- Escrever um programa que o usuário digita um número e o programa imprimi os números pares
num_pares = int(input("Digite um número para imprimir os pares até ele: "))
for i in range(1, num_pares + 1):
  if i % 2 == 0:
    print(i)
```

```python
# 5- Escrever um programa que o usuário digita um número e o programa imprimi os números impares
num_impares = int(input("Digite um número para imprimir os ímpares até ele: "))
for i in range(1, num_impares + 1):
  if i % 2 != 0:
    print(i)
```

```python
# 6- Escrever um programa que o usuário digita um número e o programa soma os números ate o número digitado
num_soma = int(input("Digite um número para somar até ele: "))
soma_total = 0
for i in range(1, num_soma + 1):
  soma_total += i
print(f"A soma dos números até {num_soma} é {soma_total}")
```

```python
# 7- Escrever um programa que o usúario digita uma palavra e o programa conta quantas letras tem a palavra
palavra = input("Digite uma palavra: ")
numero_letras = len(palavra)
print(f"A palavra '{palavra}' tem {numero_letras} letras.")
```

```python
# 8- Escrever um programa que o usuário digita uma palavra e o programa imprimi ela invertida
palavra_original = input("Digite uma palavra para inverter: ")
palavra_invertida = palavra_original[::-1]
print(f"A palavra invertida é: {palavra_invertida}")
```

```python
# 9- Escrever um programa que o usuário digita 04 números e o programa imprimi o maior
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
num3 = int(input("Digite o terceiro número: "))
num4 = int(input("Digite o quarto número: "))

maior_numero = max(num1, num2, num3, num4)

print(f"O maior número digitado é: {maior_numero}")
```

```python
# 10- Pirâmide numérica simétrica
# Faça um programa que Receba um número N e imprima um pirâmide perfeitamente centralizada

N = int(input("Digite um número inteiro N para construir a pirâmide: "))

if N <= 0:
    print("Por favor, digite um número inteiro positivo.")
else:
    for i in range(1, N + 1):

        spaces = "  " * (N - i)


        left_side = " ".join(str(j) for j in range(1, i + 1))


        right_side = " ".join(str(j) for j in range(i - 1, 0, -1))

        if right_side:
            print(f"{spaces}{left_side} {right_side}")
        else:
            print(f"{spaces}{left_side}")
```
